```python
import matplotlib.pyplot as plt
hours=[2,3,4,5,6,7,8]
marks=[50,54,65,85,94,87,76]
plt.scatter(hours,marks,color="red",marker="*",s=100,label="students")
plt.title("Hours studied vs marks obtained")
plt.xlabel("Hours studied")
plt.ylabel("Marks obtained")
plt.grid(True)
plt.legend()
plt.show
       
```

```python
import matplotlib.pyplot as plt
hours=[2,3,4,5,6,7,8]
marks=[50,54,65,85,94,87,76]
plt.scatter(hours,marks,color="blue",marker=".",s=100,label="students")
plt.title("Hours studied vs marks obtained")
plt.xlabel("Hours studied")
plt.ylabel("Marks obtained")
plt.plot(hours,marks,color="red",label="lineplot")
plt.grid(True)
plt.legend()
plt.show
       
```
